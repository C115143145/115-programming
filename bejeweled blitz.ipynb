```python
def get_gem_position(basex, basey, gap, r, c):
    """
    計算指定行列寶石的中心點座標。
    
    參數:
    basex (float): 左上角 (第0行, 第0列) 寶石的中心 X 座標
    basey (float): 左上角 (第0行, 第0列) 寶石的中心 Y 座標
    gap (float): 寶石之間的間距
    r (int): 目標寶石的行 (Row, 0~7)
    c (int): 目標寶石的列 (Column, 0~7)
    
    回傳:
    tuple: (x, y) 目標寶石的中心座標
    """
    
    # 因為 basex, basey 已經是第0行第0列的位置
    # 往右移動 c 次，每次移動 gap
    x = basex + c * gap
    
    # 往下移動 r 次，每次移動 gap
    y = basey + r * gap
    
    return (x, y)

# --- 使用範例 ---
# 假設 basex = 50, basey = 100, gap = 30
# 找尋第 2 行 (r=2) 第 3 列 (c=3) 的寶石位置
# x = 50 + 3 * 30 = 140
# y = 100 + 2 * 30 = 160
# print(get_gem_position(50, 100, 30, 2, 3)) 
# 輸出: (140, 160)
```
